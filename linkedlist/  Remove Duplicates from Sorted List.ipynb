Given the head of a sorted linked list, delete all duplicates such that each element appears only once. Return the linked list sorted as well.

 

Example 1:


Input: head = [1,1,2]
Output: [1,2]
Example 2:


Input: head = [1,1,2,3,3]
Output: [1,2,3]
 

Constraints:

The number of nodes in the list is in the range [0, 300].
-100 <= Node.val <= 100
The list is guaranteed to be sorted in ascending order.

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def deleteDuplicates(self, head: ListNode | None) -> ListNode | None:
        if head is None:
            return head
        currNode = head
        nextNode = head.next
        while nextNode and nextNode.next:
            if currNode.val == nextNode.val:
                currNode.next = nextNode.next
                nextNode = nextNode.next
                continue
            currNode = currNode.next
            nextNode = nextNode.next
        # end of list check
        if nextNode and currNode.val == nextNode.val:
                print('here')
                currNode.next = None
        return head
```

```python
head = ListNode(1)
secondNode = ListNode(1)
thirdNode = ListNode(2)
head.next = secondNode
secondNode.next = thirdNode

sol = Solution()
sortedListWithoutDup = sol.deleteDuplicates(head)
while(sortedListWithoutDup):
    print(sortedListWithoutDup.val)
    sortedListWithoutDup = sortedListWithoutDup.next
```
