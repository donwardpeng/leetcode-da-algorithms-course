## Syntax to remember



```python
# Set multiple values or 2 pointers in tandem

m = 2
n = 5
while m > 1:
    m,n = m-1, n-1
print (f'm = {m}, n = {n}')
```

### Common Stack Operations

```python
# Declaration: we will just use a list
stack = []

# Pushing elements:
stack.append(1)
stack.append(2)
stack.append(3)

# Popping elements:
stack.pop() # 3
stack.pop() # 2

# Check if empty
not stack # False

# Check element at top
stack[-1] # 1

# Get size
len(stack) # 1
```
