## Contiguous Array

Solution
Given a binary array nums, return the maximum length of a contiguous subarray with an equal number of 0 and 1.

 

Example 1:

Input: nums = [0,1]
Output: 2
Explanation: [0, 1] is the longest contiguous subarray with an equal number of 0 and 1.
Example 2:

Input: nums = [0,1,0]
Output: 2
Explanation: [0, 1] (or [1, 0]) is a longest contiguous subarray with equal number of 0 and 1.
Example 3:

Input: nums = [0,1,1,1,1,1,0,0,0]
Output: 6
Explanation: [1,1,1,0,0,0] is the longest contiguous subarray with equal number of 0 and 1.
 

Constraints:

1 <= nums.length <= 105
nums[i] is either 0 or 1.

```python
from typing import List
from collections import defaultdict
class Solution:
    def myOrig_findMaxLength(self, nums: List[int]) -> int:
        left = ans = right = 0
        counts = defaultdict(int)
        counts[(nums[0])] += 1
        for i in range(1,len(nums)):
            counts[nums[i]] += 1
            print(f'***i = {i}, counts = {counts} ')
            while(i%2 != 0 and counts[0] != counts[1] and left <= i):
                print(f'decrementing {nums[left]}')
                counts[nums[left]] -= 1
                left += 1   
                print(f'left = {left}')
                print(f'i = {i}, counts = {counts} ')
            
    def findMaxLength(self, nums: List[int]) -> int:
        maxDist = curr = 0
        # maintain the first unique count value in a hashmap - 0 adds -1, 1 adds 1
        # key = count value, value = index found
        counts = defaultdict(int) 
        # initialize first count
        counts_key_for_first_pos = -1 if nums[0]==0 else 1
        counts[counts_key_for_first_pos] = 0
        for i in range(1,len(nums)):
            if (nums[i]) == 0:
                curr -= 1
            elif (nums[i]) == 1:
                curr += 1
            if curr not in counts:
                counts[curr] = i
        print(counts)
```

```python
nums = [0,1,1,1,1,1,0,0,0]
sol = Solution()
sol.findMaxLength(nums)
```
