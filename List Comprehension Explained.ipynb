A list comprehension is a shorter, faster way to create a new list from an existing collection or loop. Think of it as a standard for loop squeezed into a single line of brackets [].

## The Golden Formula
The easiest way to remember the structure is to break it down into three parts: What you want, Where it comes from, and (optionally) The condition.

[ expression for item in iterable if condition ]


   1. expression (What you want) – The final item or modification you want to put into the new list.
   2. for item in iterable (Where it comes from) – The exact same loop syntax you use in a normal loop.
   3. if condition (The filter - optional) – Only keeps items that evaluate to True.

------------------------------

## How to Map a Loop to a Comprehension
If you can write a standard for loop, you can write a list comprehension. You just move the pieces around.

## Example: Squaring numbers
The traditional way:

```squares = []
for num in range(1, 6):
  squares.append(num * num) # 1. What you want
```

The list comprehension way:
You take what you want and place it at the very front:

```squares = [num * num for num in range(1, 6)]```

## Example with a Filter (if statement)
If you add an if statement, it simply tags onto the very end.
The traditional way:

```evens = []
for num in range(1, 6):
    if num % 2 == 0:        # 3. The condition
        evens.append(num)   # 1. What you want
```
The list comprehension way:

```evens = [num for num in range(1, 6) if num % 2 == 0]```

------------------------------
## A Mental Trick to Remember the Order
Read it like a sentence from left to right:

"Give me X, for every Y inside of Z, but only if Condition is true."


* [k for k, v in data.items() if v == 15]
* "Give me the key, for every key, value pair in my dictionary, but only if the value equals 15."
