Example 3: 844. Backspace String Compare

Given two strings s and t, return true if they are equal when both are typed into empty text editors. '#' means a backspace character.

For example, given s = "ab#c" and t = "ad#c", return true. Because of the backspace, the strings are both equal to "ac".

```python
class Solution:
    def backSpaceCompare(self, s1: str, s2: str) -> bool:
        stack1 = []
        stack2 = []
        for c in s1:
            if stack1 and c == "#":
                stack1.pop()
            else:
                stack1.append(c)
        for c in s2:
            if stack2 and c == "#":
                stack2.pop()
            else:
                stack2.append(c)
        return stack1 == stack2
        
```

```python
sol = Solution()
s1 = "ab#c"
s2 = "ad#c"
sol.backSpaceCompare(s1,s2)
```

```python
sol = Solution()
s1 = "ab##c"
s2 = "ad#c"
sol.backSpaceCompare(s1,s2)
```

```python
sol = Solution()
s1 = "ab#c"
s2 = "ad#d#c"
sol.backSpaceCompare(s1,s2)
```
